**Data analysis and preprocessing**

```python
# Data preprocessing with Numpy & Pandas

import pandas as pd

dataset = pd.read_csv("amazon_reviews_us_Apparel_v1_00.csv")

dataset['review_body'] = dataset['review_body'].dropna()
dataset['star_rating'] = dataset['star_rating'].dropna()
dataset['verified_purchase'] = dataset['verified_purchase'].dropna()

dataset['review_date'] = pd.to_datetime(dataset['review_date'])

print(dataset['review_date'])

import re

def clean_review(text):
    if not isinstance(text, str):
        return text
    text = re.sub(r'<[^>]+>', ' ', text)
    text = text.replace('&amp;', '&').replace('&lt;', '<').replace('&gt;', '>')
    text = text.replace('&quot;', '"').replace('&#39;', "'").replace('&nbsp;', ' ')
    text = re.sub(r'http\S+|www\.\S+', '', text)
    text = re.sub(r'[^a-zA-Z0-9\s.,!?\'"-]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

dataset['review_body'] = dataset['review_body'].apply(clean_review)

dataset = dataset['review_body'].dropna()

```
